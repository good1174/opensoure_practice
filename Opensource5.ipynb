```python
project = ("Open Track", "1.0", 2024, "Python", "Hongcheol")
print("==========================================")
print(f"   {project[0]} — Open Source Project")
print(f"   Project Lead :{project[4]}")
print("==========================================")
print(f"Name : {project[0]} {'|'} Version : {project[1]} {'|'} Started : {project[2]}")
print(f"First 3 fields : {project[:3]}")
print(f"Language count : {project.count('Python')} Language index : {project.index('Python')}")
#project[0] = 'test' → TypeError: tuple does not support item assignment # It is because tuple cannot be changed. But list can be changed.
print("------------------------------------------")

contributors = []
for i in range(4):
    name = input("name: ")
    role = input("role: ")
    language = input("language: ")
    commits = input("commits: ")
    country = input("country: ")
    contributors.append({
        'name': name,
        'role': role,
        'language': language,
        'commits': commits,
        'country': country
    })

names = [i['name'] for i in contributors]
names.sort()
print(f"\nSorted names : {names}")
print(f"Last name : {names[-1]}  First two : {names[:2]}")

for i in contributors:
    i.update({'status': 'Active'})

print(f"Contributor 1 status : {contributors[0].get('status')}")
backup= contributors[0].copy()
print(f"Backup : {backup}")


```

```python
issues = []
for i in range(5):
    issue_id = input("id : ")
    title = input("title : ")
    bug_type = input("type : (Bug/Feature)")
    priority = input("priority: (Critical/High/Medium/Low):")
    reporter = input("reporter : ")
    status = input("status : (Open/In Progress/Resolved): ")
    issues.append({
        'id': issue_id, 'title': title, 'type': bug_type,
        'priority': priority, 'reporter': reporter, 'status': status
    })

open_count = 0
for i in issues:
    if i['status'] == 'Open':
        open_count += 1
print(f"Open issues : {open_count}")

issues[0]['priority'] = 'Critical'
print(f"First issue → priority updated to Critical.")
print(f"Last two issues : {issues[-2:]}")

reporters = set()
for i in issues:
    reporters.add(i['reporter'])

tech_stack = set()
for i in contributors:
    tech_stack.add(i['language'])

tech_stack.add('Rust')
tech_stack.discard('R')

union_sets = reporters.union(tech_stack)
intersection_sets = reporters.intersection(tech_stack)
difference_sets = reporters.difference(tech_stack)

print('------------------------------------------')
print(f'reporters : {reporters}')
print(f'tech_stack : {tech_stack}')
print(f'union : {union_sets}')
print(f'intersection : {intersection_sets}    difference : {difference_sets}')

priority_set = {i['priority'] for i in issues}
if 'Critical' in priority_set:
    print("Critical present : YES — flag for immediate review.")
else:
    print("Critical present : No")


priority_count = {}
for i in issues:
    j = i['priority']
    priority_count[j] = priority_count.get(j, 0) + 1

status_groups = {}
for i in issues:
    j = i['status']
    if j not in status_groups:
        status_groups[j] = []
    status_groups[j].append(i['title'])

print('------------------------------------------')
print(f"priority keys : {priority_count.keys()}")
print(f"priority values : {priority_count.values()}")

status_output = " | ".join([f"{status} : {titles}" for status, titles in status_groups.items()])

print(f"Status Groups -> {status_output}")

counts = {}
for i in issues:
    counts[i['reporter']] = counts.get(i['reporter'], 0) + 1

top_reporter = ""
max = -1
for i, count in counts.items():
    if count > max:
        max = count
        top_reporter = i
print(f"Top reporter : {top_reporter} ({max} issues)")

first_thing = issues[0].pop('type')
print(f"After pop('type') : {issues[0]}")
```

```python
from logging import critical
import os

folder_name = project[0].lower().replace(' ','_')
if not os.path.exists(folder_name):
    os.makedirs(folder_name)

report_path = os.path.join(folder_name, "project_report.txt")
csv_path = os.path.join(folder_name, "issues.csv")

try:
    with open(report_path, "w") as f:
        f.write(f"==========================================\n")
        f.write(f"    {project[0]} — PROJECT REPORT\n")
        f.write(f"   Project Lead :{project[4]}\n")
        f.write(f"==========================================\n")
        f.write(f"Name : {project[0]} {'|'} Version : {project[1]} {'|'} Started : {project[2]}\n")
        f.write(f"contributors : {contributors}\n")
        f.write(f"issues : {issues}\n")
        f.write(f"priority values : {priority_count.values()}\n")
        f.write(f"Status Groups -> {status_output}\n")
        f.write(f"Top reporter : {top_reporter} ({max} issues)\n")

    with open(csv_path, "w") as f:
        f.write("id,title,priority,reporter,status\n")
        for i in issues:
            f.write(f"{i['id']},{i['title']},{i['priority']},{i['reporter']},{i['status']}\n")
    print(f"Folder created : {folder_name}/")
    print(f"Files saved : {os.listdir(folder_name)}")
except IOError as e:
    print(f"{e}")

try:
    print("--- read() ---")
    with open(report_path, "r") as f:
        print(f.read())

    print("--- readline() ---")
    with open(report_path, "r") as f:
        print(f"Line 1 : {f.readline().strip()}")
        print(f"Line 2 : {f.readline().strip()}")

    print("--- readlines() ---")
    with open(report_path, "r") as f:
        lines = f.readlines()
        critical_high_lines = [i.strip() for i in lines if "Critical" in i or "High" in i]
        print(f"Total lines : {len(lines)}    Critical/High lines found : {len(critical_high_lines)}")
except FileNotFoundError as e:
    print(f"{e}")

print('==========================================')
print(f"    {project[0]} — FINAL SUMMARY")
print('==========================================')
print(f"Project : {project[0]}    Version : {project[1]}    Lead : {project[4]}")
print(f"Contributors : {len(contributors)}    Names : {names}")
print(f"Tech Stack : {tech_stack}")
print(f"Issues : {len(issues)}    Open : {open_count}   Reporters : {len(reporters)}")
print(f"Top Reporter : {top_reporter} ({max} issues)")
print(*[f"{k}:{v}" for k, v in priority_count.items()])
print(f"Report : {report_path}")
print(f"CSV : {csv_path}")
print('==========================================')
print(f"{project[0]} complete. Thank you for contributing to open source!")
print('==========================================')
```
